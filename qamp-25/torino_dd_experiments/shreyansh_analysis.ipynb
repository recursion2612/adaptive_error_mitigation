```python
## Reading data 
import pandas as pd

backend = pd.read_csv("Backend_Shreyansh.csv")
simulator = pd.read_csv("Simulator_Shreyansh.csv")
```

```python
backend["DD_works"] = backend["hellinger_fidelity_no_dd"] <= backend["hellinger_fidelity_dd"]
simulator["DD_works"] = simulator["hellinger_fidelity_no_dd"] <= simulator["hellinger_fidelity_dd"]
```

```python
import matplotlib.pyplot as plt
data = [simulator["sparsity_by_layer"], simulator["sparsity_by_time"], simulator[simulator["DD_works"] == True]["sparsity_by_layer"], simulator[simulator["DD_works"] == True]["sparsity_by_time"]]



labels = ['sparsity_by_layer', 'sparsity_by_time', 'DD_works: sparsity_by_layer', 'DD_works: sparsity_by_time']
colors = ['cyan', 'steelblue', 'cyan', 'steelblue']

fig, ax = plt.subplots(figsize=(12, 5))
ax.set_ylabel('Magnitude')
ax.set_title("Dist. of sparsity metrics")

bplot = ax.boxplot(data,
                   patch_artist=True,  # fill with color
                   tick_labels=labels,
                   meanline=True, medianprops=dict(color='red'))  # will be used to label x-ticks

# fill with colors
for patch, color in zip(bplot['boxes'], colors):
    patch.set_facecolor(color)
plt.grid(True, axis='y')
plt.show()
```

```python
point_one_sbl           = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.1]["sparsity_by_layer"]
point_zero_one_sbl      = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.01]["sparsity_by_layer"]
point_zero_zero_one_sbl = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.001]["sparsity_by_layer"]

point_one_sbt           = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.1]["sparsity_by_time"]
point_zero_one_sbt      = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.01]["sparsity_by_time"]
point_zero_zero_one_sbt = backend[backend["hellinger_fidelity_no_dd"] - backend["hellinger_fidelity_dd"]<=0.001]["sparsity_by_time"]


data_sbl = [backend["sparsity_by_layer"], point_one_sbl, point_zero_one_sbl, point_zero_zero_one_sbl]

data_sbt = [backend["sparsity_by_time"], point_one_sbt, point_zero_one_sbt, point_zero_zero_one_sbt]


labels_sbl = ['sparsity_by_layer', 'hf_diff=0.1', 'hf_diff=0.01', 'hf_diff=0.001']
labels_sbt = ['sparsity_by_time', 'hf_diff=0.1', 'hf_diff=0.01', 'hf_diff=0.001']

colors_sbl = ['cyan', 'cyan', 'cyan', 'cyan']
colors_sbt = ['steelblue', 'steelblue', 'steelblue', 'steelblue']


fig, ax = plt.subplots(nrows=2, figsize=(12, 10))
ax[0].set_ylabel('Magnitude')
ax[0].set_title("Dist. of sparsity by layer")

ax[1].set_ylabel('Magnitude')
ax[1].set_title("Dist. of sparsity by time")

sbl = ax[0].boxplot(data_sbl,
                   patch_artist=True,  # fill with color
                   tick_labels=labels_sbl,
                   meanline=True, medianprops=dict(color='red'))  # will be used to label x-ticks
sbt = ax[1].boxplot(data_sbt,
                   patch_artist=True,  # fill with color
                   tick_labels=labels_sbt,
                   meanline=True, medianprops=dict(color='red'))  # will be used to label x-ticks
# fill with colors
for patch, color in zip(sbl['boxes'], colors_sbl):
    patch.set_facecolor(color)

for patch, color in zip(sbt['boxes'], colors_sbt):
    patch.set_facecolor(color)



ax[0].grid(True, axis='y')
ax[1].grid(True, axis='y')

plt.show()
```

```python
import ast
my_data = pd.read_csv("GHZ-on-hardware.csv")

my_data.ideal_counts = my_data.ideal_counts.apply(ast.literal_eval)
my_data.simulated_counts = my_data.simulated_counts.apply(ast.literal_eval)
my_data.no_dd_counts = my_data.no_dd_counts.apply(ast.literal_eval)
my_data.dd_counts = my_data.dd_counts.apply(ast.literal_eval)
my_data["hellinger_fidelity(ideal)"] = my_data["hellinger_fidelity(ideal)"].apply(ast.literal_eval)
my_data["hellinger_fidelity(simulated)"] = my_data["hellinger_fidelity(simulated)"].apply(ast.literal_eval)



reduced_data = my_data.drop(["ideal_counts", "simulated_counts", "hellinger_fidelity(ideal)", "hellinger_fidelity(simulated)"], axis=1)




reduced_data["no_dd_fidelity"] = [fidelity[1] for fidelity in my_data["hellinger_fidelity(ideal)"]]
reduced_data["dd_fidelity"] = [fidelity[2] for fidelity in my_data["hellinger_fidelity(ideal)"]]
reduced_data
```
